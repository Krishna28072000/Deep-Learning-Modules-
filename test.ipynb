```python
import tensorflow as tf

tensor = tf.constant([[1,2,3],[4,5,6]])
rank = tf.ransk(tensor)
shape = tf.shape(tensor)
size = tf.size(tensor)
print("TENSOR", tensor)
print("")
print("")
print("")
print("TENSOR SHAPE", shape)
print("")
print("")
print("")
print("TENSOR SIZE", size)
print("")
print("")
print("")
print("TENSOR RANK", rank)
```

```python
t2 = tf.constant([[0,1,2],[3,4,5],[6,7,8]])
t2
```

```python
first_row = tf.slice(t2, [0,0] , [1, tf.shape(t2)[1]])
print("FIRST ROW",first_row)
```

```python
second_column = tf.slice(t2, [0,1], [tf.shape(t2)[0], 1])
second_column
```

```python
second_and_third_columns = tf.slice(t2, [0,1], [tf.shape(t2)[0], 2])
second_and_third_columns
```

```python
import tensorflow as tf
t = tf.constant([[1,2,3],
                [4,5,6]])
t
```

```python
result = tf.reshape(t, [-1])
print("The result is :\n",result)
print('\n Original Tensor Shape:',tf.shape(t))  
print('Flattened Tensor Shape:',tf.shape(result))  
```
